```python
# Function to read the Project Info sheet from each workbook
# Returns a dictionary of key-value pairs from columns A and B
def read_project_info(workbook_path):
    
    # Load workbook with data_only=True to get calculated values not formulas
    wb = openpyxl.load_workbook(workbook_path, data_only=True)
    
    # Select the Project Info sheet
    ws = wb["Project Info"]
    
    # Iterate through rows 1 to 9, columns A and B
    # Column A is the field name, Column B is the value
    info = {}
    for row in ws.iter_rows(min_row=1, max_row=10, min_col=1, max_col=2):
        key = row[0].value
        value = row[1].value
        if key:
            info[key] = value
    
    return info

# Test the function on the NHS hospital workbook
# This confirms the function reads the correct fields before we loop all five
info = read_project_info(workbooks["NHS Acute Hospital Wing"])
for k, v in info.items():
    print(f"{k}: {v}")
```
